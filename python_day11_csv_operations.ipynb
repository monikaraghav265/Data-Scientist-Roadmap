```python
import pandas as pd

data = {
    "Name": ["Aman", "Riya", "Rahul", "Sneha", "Karan"],
    "Age": [21, 22, 20, 23, 24],
    "Marks": [85, 90, 78, 88, 92]
}

df = pd.DataFrame(data)
```

```python
df.to_csv("students.csv", index=False)     
```

```python
df2 = pd.read_csv("students.csv")  
df2
```

```python
df2.head() 
```

```python
df2.info()
df2.describe()
```

```python
df2[df2["Marks"] > 85]
df2.sort_values(by="Marks", ascending=False)
```

```python
#Checking missing values 
df2.isnull().sum()
```

```python
#Add New Column
df2["Status"] = df2["Marks"].apply(
    lambda x: "Excellent" if x > 85 else "Average"
)
df2["Status"]
```

```python
df2.rename(columns={"Marks": "Score"}, inplace=True)
```

```python
df2.drop("Age", axis=1, inplace=True)
```

```python
df2.sort_values(by="Score", ascending=False)
```

```python
df2["Status"].value_counts()
```
