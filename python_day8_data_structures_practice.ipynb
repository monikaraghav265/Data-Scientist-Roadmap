Count Word Frequency 

```python
text = "data science is fun and data science is powerful"
words = text.split()
print(words)
```

```python
text = "data science is fun and data science is powerful"

words = text.split()

freq = {}

for word in words:
    freq[word] = freq.get(word, 0) + 1

print(freq)
```

Find Largest Number 

```python
nums = [10, 45, 23, 89, 12]

largest = nums[0]

for num in nums:
    if num > largest:
        largest = num

print(largest)

# Used a loop to compare each element and find the largest number in the list.
```

Character Frequency in string 

```python
s = "datascience"

char_count = {}

for ch in s:
    char_count[ch] = char_count.get(ch, 0) + 1

print(char_count)

# Used a dictionary to count the frequency of each character in the string.
```
