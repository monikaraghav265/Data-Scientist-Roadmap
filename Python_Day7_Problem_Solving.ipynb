**Problem: Most Frequent Character**

```python
s="data science"
freq={}
for ch in s:
    if ch in freq:
        freq[ch]+=1
    else:
        freq[ch]=1
print(freq)


```

```python
s="data science"
max_char=None
max_count=0
for key in s:
    if freq[key]>max_count:
        max_count=freq[key]
        max_char=key
print(max_char)
```

**Remove Duplicates**

```python
lst=[1, 2, 2, 3, 4, 1, 5]    
result=list(set(lst))     #Here order not maintain bcz set is an unordered data structure 
print(result)
```

**Remove Duplicates(Order Maintain)**

```python
lst=[1, 2, 2, 3, 4, 1, 5] 
result=list(dict.fromkeys(lst))      #Maintain order 
print(result) 
```

**Group Words by Length**

```python
words=["cat", "dog", "elephant", "bat", "tiger"]
grouped={}
for word in words:
    length=len(word)
    if length in grouped:
        grouped[length].append(word)
    else:
        grouped[length]=[word]
print(grouped) 
        
    
```
