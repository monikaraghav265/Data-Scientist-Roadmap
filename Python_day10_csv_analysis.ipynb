Focus: Reading and analyzing real dataset using Pandas

```python
import pandas as pd

data = {
    "Name": ["Aman", "Riya", "Rahul", "Sneha", "Karan"],
    "Age": [21, 22, 20, 23, 24],
    "Marks": [85, 90, 78, 88, 92]
}

df = pd.DataFrame(data)
```

```python
df.to_csv("data.csv", index=False) 
```
