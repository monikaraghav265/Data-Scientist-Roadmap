```python
#Word frequency counter 
text="apple mango banana apple apple banana"
words=text.split()
freq={}
for word in words:
    if word in freq:
        freq[word]+=1
    else:
        freq[word]=1
print(freq)
```

```python
#Count elements in list using dictionary 
lst=[1,2,2,3,4,4,5,5,5,6]
res={}
for elm in lst:
    res[elm]=lst.count(elm)
print(res)
    

```

```python
#union and intersection   #set opetations 
a={1,2,3,4,5}
b={4,5,6,7,8}
print(a.union(b))
print(a.intersection(b)) 
```

```python
#frequency count using dictionary
lst=[1,2,2,3,4,4,5,6,6]
frq={}
for i in lst:
    if i in frq:
        frq[i]+=1
    else:
        frq[i]=1
print(frq) 

```

```python
#First non repeating character 
s="aaabbbcde"
for ch in s:
    if s.count(ch)==1:
        print(ch)
        break
    
```

```python
#List---Dictionary(index as value) 
lst=['a','b','c','d']
result={}
i=0
for item in lst:
    result[item]=i
    i+=1
print(result) 
```

```python
#Common elements in set 
a={1,2,3,4,5}
b={5,6,2,7,8}
print(a.intersection(b)) 
```
