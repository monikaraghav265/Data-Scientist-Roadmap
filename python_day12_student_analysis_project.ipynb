Day 12 – Student Data Analysis Project
Focus: Data Analysis using Pandas
Goal: Extract insights from dataset

```python
import pandas as pd

data = {
    "Name": ["Aman", "Riya", "Rahul", "Sneha", "Karan", "Neha"],
    "Age": [21, 22, 20, 23, 24, 21],
    "Marks": [85, 90, 78, 88, 92, 75]
}

df = pd.DataFrame(data)
df
```

```python
df["Grade"] = df["Marks"].apply(
    lambda x: "A" if x >= 85 else "B"
)
df["Grade"]
```

```python
df["Marks"].max()
```

```python
df["Marks"].mean()
```

```python
df[df["Marks"] > df["Marks"].mean()]
```

```python
df.groupby("Grade")["Marks"].mean()
```

```python
df.sort_values(by="Marks", ascending=False)
```

```python
import matplotlib.pyplot as plt

df["Marks"].plot(kind="bar")
plt.title("Student Marks")
plt.show()
```

Conclusion:
- Students with Grade A have higher average marks
- Majority of students scored above 80
- Data analysis helps identify performance trends
