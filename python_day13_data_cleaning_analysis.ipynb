Day 13 – Data Cleaning and Advanced Analysis

Focus: Handling missing data and improving dataset quality

```python
#Dataset create
import pandas as pd
import numpy as np

data = {
    "Name": ["Aman", "Riya", "Rahul", "Sneha", "Karan", "Neha"],
    "Age": [21, 22, np.nan, 23, 24, 21],
    "Marks": [85, 90, 78, np.nan, 92, 75]
}

df = pd.DataFrame(data)
df
```

```python
#Check Missing Values 
df.isnull().sum()
```

```python
#Handle Missing Values 
df['Age'].fillna(df['Age'].mean(),inplace=True)
```

```python
df['Marks'].fillna(df['Marks'].mean(),inplace=True) 
```

```python
df
```

```python
df.isnull().sum()
```

```python
# Add Category Column 
df["Performance"] = df["Marks"].apply(
    lambda x: "Good" if x >= 85 else "Average"
)
df
```

```python
# Group Analysis
df.groupby("Performance")["Marks"].mean()
```

```python
# Sorting
df.sort_values(by="Marks", ascending=False)
```

```python
# Visualization
import matplotlib.pyplot as plt

df["Marks"].plot(kind="bar")
plt.title("Marks Distribution After Cleaning")
plt.show() 
```

Conclusion:
- Missing values were handled using mean
- Data became clean and usable
- Performance categories help analyze student results
